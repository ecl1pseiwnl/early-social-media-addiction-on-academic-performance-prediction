```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
import os
import warnings as wr
wr.filterwarnings('ignore')

dframe = pd.read_csv('scaling-dataset.csv')
print(dframe.head())

output_dir = "Asset/"
os.makedirs(output_dir,exist_ok=True)

#Scaling data
pos_cols = ["Daily_Usage_Hours", "Sleep_Duration_Hours", "Sleep_Quality_Score", "Mental_Health_Index", "Academic_Performance_GPA"]
zero_cols = ["Perceived_Stress_Score","Weekend_Extra_Hours"]

dframe_scaled = dframe.copy()

print("Processing data scaling...")
#Apply Box-Cox
for col in pos_cols:
    transform_data, best_lambda = stats.boxcox(dframe[col])
    dframe_scaled[col+'_BoxCox'] = transform_data
    print(f"[{col}] Box-Cox transformation applied (Optimal Lambda = {best_lambda:.4f})")
    
#Apply Log1p
for col in zero_cols:
    dframe_scaled[col + '_Log1p'] = np.log1p(dframe[col])
    print(f"[{col}] Log1p transformation applied")

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
sns.set_theme(style="whitegrid")

# Daily Usage Hour (Box-Cox)
sns.histplot(dframe['Daily_Usage_Hours'], kde=True, ax=axes[0, 0], color='#4C72B0')
axes[0, 0].set_title('Original: Daily_Usage_Hours', fontsize=12, fontweight='bold')

sns.histplot(dframe_scaled['Daily_Usage_Hours_BoxCox'], kde=True, ax=axes[0, 1], color='#55A868')
axes[0, 1].set_title('Scaled (Box-Cox): Daily_Usage_Hours', fontsize=12, fontweight='bold')

# Perceived Stress Score (Log1p)
sns.histplot(dframe['Perceived_Stress_Score'], kde=True, ax=axes[1, 0], color='#C44E52')
axes[1, 0].set_title('Original: Perceived Stress Score', fontsize=12, fontweight='bold')

sns.histplot(dframe_scaled['Perceived_Stress_Score_Log1p'], kde=True, ax=axes[1, 1], color='#8172B2')
axes[1, 1].set_title('Scaled (Log1p): Perceived Stress Score', fontsize=12, fontweight='bold')

plt.suptitle('Data Distribution Before and After Transformation', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()

plot_path = os.path.join(output_dir, 'scaling_transformation_EDA.png')
plt.savefig(plot_path, dpi=300, bbox_inches='tight')
print(f"\n![Transformation Plot saved at]({plot_path})")
plt.close() 

filename_output = "scaled-dataset.csv"
dframe_scaled.to_csv(filename_output,index=False)
print(f"\nScaled dataset successfully save at {filename_output}")

cols_to_drop = [
    'Daily_Usage_Hours', 
    'Sleep_Duration_Hours', 
    'Sleep_Quality_Score', 
    'Mental_Health_Index',  
    'Perceived_Stress_Score',
    'Weekend_Extra_Hours',
    'Academic_Performance_GPA'
]
dframe_final = dframe_scaled.drop(columns=cols_to_drop)

y = dframe_final['Overall_Impact']
x = dframe_final.drop(columns=['Overall_Impact'])

print("\n--- X AND y SPLIT RESULTS ---")
print("Features (X) shape:", x.shape)
print("Target (y) shape:", y.shape)
print("\nFinal features included in the model:")
print(X.columns.tolist())
```
